# Sistema de Roteamento de Operações para Filas por Setor

## Objetivo
Consultar operações pendentes da tabela `OPERACAO_PO` no Supabase, extrair a operação e seu setor correspondente (MT1, MT2, MT3, etc.) da coluna `operacao_pend`, e rotear automaticamente essas operações para as filas corretas no site de Gestão PPCP.

## Fluxo de Dados
1. Query na tabela `OPERACAO_PO` → coluna `operacao_pend`
2. Extrair operação e setor reference
3. Mapear operação → setor correto
4. Enviar para fila do setor no website
5. Atualizar display em tempo real

```python
# 1. IMPORTAR BIBLIOTECAS NECESSÁRIAS
import pandas as pd
import requests
import json
import logging
from datetime import datetime
from typing import Dict, List, Optional, Tuple

# Supabase
from supabase import create_client, Client

# Configurar logging
logging.basicConfig(
    level=logging.INFO,
    format='%(asctime)s - %(levelname)s - %(message)s',
    handlers=[
        logging.StreamHandler(),
        logging.FileHandler('operacoes_routing.log')
    ]
)
logger = logging.getLogger(__name__)

print("✓ Bibliotecas importadas com sucesso")
```

```python
# 3. CONSULTAR TABELA DE OPERAÇÕES PENDENTES

def query_operacoes_pendentes() -> Optional[pd.DataFrame]:
    """
    Consulta a tabela OPERACAO_PO e retorna operações pendentes.
    
    Returns:
        DataFrame com operações pendentes ou None em caso de erro
    """
    try:
        # Query na tabela OPERACAO_PO
        response = supabase.table('OPERACAO_PO').select(
            'id, operacao_pend, setor, created_at'
        ).eq('status', 'pendente').execute()  # Ajuste o filtro conforme sua estrutura
        
        if response.data:
            df = pd.DataFrame(response.data)
            logger.info(f"Total de operações pendentes encontradas: {len(df)}")
            print(f"✓ {len(df)} operações pendentes carregadas")
            return df
        else:
            logger.warning("Nenhuma operação pendente encontrada")
            print("⚠ Nenhuma operação pendente encontrada")
            return None
            
    except Exception as e:
        logger.error(f"Erro ao consultar operações pendentes: {str(e)}")
        print(f"✗ Erro na consulta: {str(e)}")
        return None

# Executar query
df_operacoes = query_operacoes_pendentes()
if df_operacoes is not None:
    print("\nPrimeiras operações encontradas:")
    print(df_operacoes.head())
```

```python
# 4. EXTRAIR OPERAÇÃO E INFORMAÇÃO DE SETOR

def extrair_operacao_e_setor(operacao_pend: str) -> Tuple[str, str]:
    """
    Extrai o nome da operação e o setor de uma string operacao_pend.
    
    Exemplo: "ACABAMENTO LAM." → ("ACABAMENTO LAM.", "LAMINAÇÃO")
             "ALINHAR MT1" → ("ALINHAR", "MT1")
    
    Args:
        operacao_pend: String com operação pendente
        
    Returns:
        Tuple (operacao, setor)
    """
    try:
        operacao_pend = operacao_pend.strip()
        
        # Verificar se contém referência de setor (MT1, MT2, MT3, etc.)
        setores_diretos = ['MT1', 'MT2', 'MT3', 'MT4', 'PINTURA', 'QUALIDADE', 
                          'LAMINAÇÃO', 'USINAGEM', 'ESTAMPARIA', 'SERV.EXT']
        
        setor_encontrado = None
        
        # Buscar setor direto na operação
        for setor in setores_diretos:
            if setor in operacao_pend.upper():
                setor_encontrado = setor
                break
        
        if setor_encontrado:
            # Remover setor do nome da operação
            operacao = operacao_pend.replace(setor_encontrado, '').strip()
            return (operacao if operacao else operacao_pend, setor_encontrado)
        else:
            # Se não encontrar setor direto, retornar como está
            logger.warning(f"Setor não identificado em: {operacao_pend}")
            return (operacao_pend, "INDEFINIDO")
            
    except Exception as e:
        logger.error(f"Erro ao extrair operação/setor de '{operacao_pend}': {str(e)}")
        return (operacao_pend, "ERRO")

# Testar extração
if df_operacoes is not None:
    print("\nExtraindo operações e setores:")
    df_operacoes[['operacao_nome', 'setor_destino']] = df_operacoes['operacao_pend'].apply(
        lambda x: pd.Series(extrair_operacao_e_setor(x))
    )
    print(df_operacoes[['operacao_pend', 'operacao_nome', 'setor_destino', 'setor']].head(10))
```

```python
# 5. MAPEAR OPERAÇÕES PARA FILAS DE SETOR

class GerenciadorFilasSetor:
    """
    Gerencia o roteamento de operações para as filas de cada setor.
    """
    
    # Mapeamento de setores para suas configurações
    SETORES_CONFIG = {
        'MT1': {'nome': 'MT1', 'url_fila': '/api/fila/mt1'},
        'MT2': {'nome': 'MT2', 'url_fila': '/api/fila/mt2'},
        'MT3': {'nome': 'MT3', 'url_fila': '/api/fila/mt3'},
        'MT4': {'nome': 'MT4', 'url_fila': '/api/fila/mt4'},
        'PINTURA': {'nome': 'PINTURA', 'url_fila': '/api/fila/pintura'},
        'QUALIDADE': {'nome': 'QUALIDADE', 'url_fila': '/api/fila/qualidade'},
        'LAMINAÇÃO': {'nome': 'LAMINAÇÃO', 'url_fila': '/api/fila/laminacao'},
        'USINAGEM': {'nome': 'USINAGEM', 'url_fila': '/api/fila/usinagem'},
        'ESTAMPARIA': {'nome': 'ESTAMPARIA', 'url_fila': '/api/fila/estamparia'},
        'SERV.EXT': {'nome': 'SERV.EXT', 'url_fila': '/api/fila/serv_ext'},
    }
    
    def __init__(self, url_base_website: str = "http://localhost"):
        """
        Inicializa o gerenciador de filas.
        
        Args:
            url_base_website: URL base do website (ex: http://localhost)
        """
        self.url_base = url_base_website
        
    def mapear_operacao_para_fila(self, operacao_id: str, operacao_nome: str, 
                                   setor: str, ordem: str, cliente: str) -> Dict:
        """
        Mapeia uma operação para a fila do setor correto.
        
        Args:
            operacao_id: ID único da operação
            operacao_nome: Nome da operação
            setor: Setor destino (MT1, MT2, etc.)
            ordem: Número da ordem
            cliente: Cliente responsável
            
        Returns:
            Dict com mapeamento da operação
        """
        try:
            # Validar setor
            if setor not in self.SETORES_CONFIG:
                logger.warning(f"Setor inválido: {setor}. Usando INDEFINIDO")
                setor = "INDEFINIDO"
            
            mapeamento = {
                'id': operacao_id,
                'operacao': operacao_nome,
                'setor': setor,
                'ordem': ordem,
                'cliente': cliente,
                'timestamp': datetime.now().isoformat(),
                'status_fila': 'aguardando',
                'url_endpoint': f"{self.url_base}{self.SETORES_CONFIG.get(setor, {}).get('url_fila', '/api/fila/indefinido')}"
            }
            
            return mapeamento
            
        except Exception as e:
            logger.error(f"Erro ao mapear operação {operacao_id}: {str(e)}")
            return None
    
    def processar_lote_operacoes(self, df: pd.DataFrame) -> List[Dict]:
        """
        Processa um lote de operações e retorna mapeamentos.
        
        Args:
            df: DataFrame com operações
            
        Returns:
            Lista de operações mapeadas
        """
        operacoes_mapeadas = []
        
        for idx, row in df.iterrows():
            try:
                setor = row.get('setor_destino', row.get('setor', 'INDEFINIDO'))
                
                mapeamento = self.mapear_operacao_para_fila(
                    operacao_id=str(row.get('id', f'OP_{idx}')),
                    operacao_nome=row.get('operacao_nome', row.get('operacao_pend', '')),
                    setor=setor,
                    ordem=str(row.get('ordem', 'N/A')),
                    cliente=str(row.get('cliente', 'N/A'))
                )
                
                if mapeamento:
                    operacoes_mapeadas.append(mapeamento)
                    
            except Exception as e:
                logger.error(f"Erro ao processar linha {idx}: {str(e)}")
                continue
        
        return operacoes_mapeadas

# Criar instância do gerenciador
gerenciador = GerenciadorFilasSetor(url_base_website="http://localhost")

# Mapear operações
if df_operacoes is not None:
    operacoes_mapeadas = gerenciador.processar_lote_operacoes(df_operacoes)
    print(f"\n✓ {len(operacoes_mapeadas)} operações mapeadas com sucesso\n")
    
    # Exibir amostra dos mapeamentos
    for op in operacoes_mapeadas[:3]:
        print(f"Operação: {op['operacao']} → Setor: {op['setor']}")
```

```python
# 6. ENVIAR OPERAÇÕES PARA FILAS DO WEBSITE

class ClienteFilasWebsite:
    """
    Comunica com o website para atualizar as filas de cada setor.
    """
    
    def __init__(self, url_base: str = "http://localhost"):
        self.url_base = url_base
        self.timeout = 5
        
    def enviar_operacao_para_fila(self, operacao: Dict, setor: str) -> bool:
        """
        Envia uma operação para a fila do setor no website.
        
        Args:
            operacao: Dict com dados da operação
            setor: Setor destino
            
        Returns:
            True se enviado com sucesso, False caso contrário
        """
        try:
            # Endpoint para adicionar à fila
            endpoint = f"{self.url_base}/api/fila/{setor.lower()}/adicionar"
            
            payload = {
                'operacao_id': operacao['id'],
                'operacao_nome': operacao['operacao'],
                'ordem': operacao['ordem'],
                'cliente': operacao['cliente'],
                'timestamp_entrada': operacao['timestamp']
            }
            
            response = requests.post(
                endpoint,
                json=payload,
                timeout=self.timeout
            )
            
            if response.status_code in [200, 201]:
                logger.info(f"Operação {operacao['id']} enviada para fila {setor}")
                return True
            else:
                logger.error(f"Erro ao enviar para {setor}: Status {response.status_code}")
                return False
                
        except requests.exceptions.ConnectionError:
            logger.error(f"Erro de conexão ao enviar para {self.url_base}")
            return False
        except Exception as e:
            logger.error(f"Erro ao enviar operação: {str(e)}")
            return False
    
    def atualizar_todas_as_filas(self, operacoes_mapeadas: List[Dict]) -> Dict:
        """
        Atualiza todas as filas de setores com operações mapeadas.
        
        Args:
            operacoes_mapeadas: Lista de operações mapeadas
            
        Returns:
            Dict com estatísticas de envio
        """
        estatisticas = {
            'total': len(operacoes_mapeadas),
            'enviadas': 0,
            'falhas': 0,
            'por_setor': {}
        }
        
        # Agrupar por setor
        operacoes_por_setor = {}
        for op in operacoes_mapeadas:
            setor = op['setor']
            if setor not in operacoes_por_setor:
                operacoes_por_setor[setor] = []
            operacoes_por_setor[setor].append(op)
        
        # Enviar para cada setor
        for setor, operacoes in operacoes_por_setor.items():
            enviadas = 0
            for op in operacoes:
                if self.enviar_operacao_para_fila(op, setor):
                    enviadas += 1
                    estatisticas['enviadas'] += 1
                else:
                    estatisticas['falhas'] += 1
            
            estatisticas['por_setor'][setor] = {
                'total': len(operacoes),
                'enviadas': enviadas,
                'falhas': len(operacoes) - enviadas
            }
        
        return estatisticas

# Criar cliente do website
cliente_website = ClienteFilasWebsite(url_base="http://localhost")

# Enviar operações (demonstração)
if operacoes_mapeadas:
    print("Enviando operações para filas de setores...\n")
    stats = cliente_website.atualizar_todas_as_filas(operacoes_mapeadas)
    
    print(f"✓ Total de operações processadas: {stats['total']}")
    print(f"✓ Operações enviadas com sucesso: {stats['enviadas']}")
    print(f"✗ Falhas ao enviar: {stats['falhas']}")
    print(f"\nResumo por setor:")
    for setor, dados in stats['por_setor'].items():
        print(f"  {setor}: {dados['enviadas']}/{dados['total']} enviadas")
```

```python
# 7. TRATAMENTO DE ERROS E LOGGING

class GestorOperacoesPipeline:
    """
    Pipeline completo para gestão de operações com tratamento robusto de erros.
    """
    
    def __init__(self, supabase_client: Client, url_website: str):
        self.supabase = supabase_client
        self.gerenciador = GerenciadorFilasSetor(url_base_website=url_website)
        self.cliente_web = ClienteFilasWebsite(url_base=url_website)
        
    def validar_dados_operacao(self, operacao: Dict) -> Tuple[bool, str]:
        """
        Valida se uma operação tem dados completos e corretos.
        
        Args:
            operacao: Dict com dados da operação
            
        Returns:
            Tuple (valido, mensagem_erro)
        """
        validacoes = {
            'operacao_pend': 'Campo operacao_pend vazio',
            'setor': 'Campo setor vazio',
            'ordem': 'Campo ordem vazio',
        }
        
        for campo, erro_msg in validacoes.items():
            if campo not in operacao or not operacao[campo]:
                return (False, erro_msg)
        
        return (True, "OK")
    
    def processar_com_tratamento_erros(self, df_operacoes: pd.DataFrame) -> Dict:
        """
        Processa operações com tratamento completo de erros.
        
        Args:
            df_operacoes: DataFrame com operações
            
        Returns:
            Dict com resultado do processamento
        """
        resultado = {
            'timestamp': datetime.now().isoformat(),
            'total_processadas': 0,
            'sucesso': 0,
            'validacao_falhou': 0,
            'mapeamento_falhou': 0,
            'envio_falhou': 0,
            'erros': [],
            'operacoes_processadas': []
        }
        
        try:
            if df_operacoes is None or df_operacoes.empty:
                resultado['erros'].append("DataFrame vazio ou None")
                logger.warning("Nenhuma operação para processar")
                return resultado
            
            # Processar cada operação
            for idx, row in df_operacoes.iterrows():
                try:
                    resultado['total_processadas'] += 1
                    
                    # 1. Validar dados
                    valido, msg_validacao = self.validar_dados_operacao(dict(row))
                    if not valido:
                        logger.warning(f"Validação falhou (linha {idx}): {msg_validacao}")
                        resultado['validacao_falhou'] += 1
                        resultado['erros'].append(f"Linha {idx}: {msg_validacao}")
                        continue
                    
                    # 2. Extrair operação e setor
                    operacao_nome, setor = extrair_operacao_e_setor(row['operacao_pend'])
                    
                    # 3. Mapear para fila
                    mapeamento = self.gerenciador.mapear_operacao_para_fila(
                        operacao_id=str(row.get('id', f'OP_{idx}')),
                        operacao_nome=operacao_nome,
                        setor=setor,
                        ordem=str(row.get('ordem', 'N/A')),
                        cliente=str(row.get('cliente', 'N/A'))
                    )
                    
                    if not mapeamento:
                        resultado['mapeamento_falhou'] += 1
                        resultado['erros'].append(f"Falha no mapeamento da linha {idx}")
                        continue
                    
                    # 4. Enviar para website
                    enviado = self.cliente_web.enviar_operacao_para_fila(
                        mapeamento, setor
                    )
                    
                    if enviado:
                        resultado['sucesso'] += 1
                        resultado['operacoes_processadas'].append({
                            'id': mapeamento['id'],
                            'operacao': mapeamento['operacao'],
                            'setor': setor,
                            'status': 'ENVIADA'
                        })
                        logger.info(f"Operação {mapeamento['id']} processada com sucesso")
                    else:
                        resultado['envio_falhou'] += 1
                        resultado['erros'].append(f"Falha ao enviar operação {mapeamento['id']}")
                        
                except Exception as e:
                    resultado['erros'].append(f"Exceção na linha {idx}: {str(e)}")
                    logger.error(f"Exceção ao processar linha {idx}: {str(e)}")
                    continue
            
            # Log resumo
            logger.info(f"Pipeline finalizado: {resultado['sucesso']}/{resultado['total_processadas']} sucesso")
            
        except Exception as e:
            resultado['erros'].append(f"Erro geral no pipeline: {str(e)}")
            logger.error(f"Erro geral no pipeline: {str(e)}")
        
        return resultado

# Executar pipeline completo
print("=" * 60)
print("EXECUTANDO PIPELINE DE GESTÃO DE OPERAÇÕES")
print("=" * 60)

pipeline = GestorOperacoesPipeline(
    supabase_client=supabase,
    url_website="http://localhost"
)

resultado_final = pipeline.processar_com_tratamento_erros(df_operacoes)

# Exibir resultado
print("\n" + "=" * 60)
print("RESULTADO FINAL DO PROCESSAMENTO")
print("=" * 60)
print(f"\nTimestamp: {resultado_final['timestamp']}")
print(f"Total de operações processadas: {resultado_final['total_processadas']}")
print(f"✓ Sucesso: {resultado_final['sucesso']}")
print(f"⚠ Validação falhou: {resultado_final['validacao_falhou']}")
print(f"⚠ Mapeamento falhou: {resultado_final['mapeamento_falhou']}")
print(f"✗ Envio falhou: {resultado_final['envio_falhou']}")

if resultado_final['operacoes_processadas']:
    print(f"\nOperações processadas com sucesso:")
    for op in resultado_final['operacoes_processadas'][:5]:
        print(f"  - {op['id']}: {op['operacao']} → {op['setor']}")

if resultado_final['erros']:
    print(f"\nErros encontrados ({len(resultado_final['erros'])}):")
    for erro in resultado_final['erros'][:5]:
        print(f"  ⚠ {erro}")
```
